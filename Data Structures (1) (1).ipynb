```python
el = []
ol = []

for i in range(1, 21):
    if i % 2 == 0:
        el.append(i)
    else:
        ol.append(i)

et = tuple(el)
ot = tuple(ol)

print("Even tuple:", et)
print("Odd tuple:", ot)
```
