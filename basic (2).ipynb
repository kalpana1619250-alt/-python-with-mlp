````python
acc_bal=float(input("enter a balance amount:"))
wd_amnt=float(input("enter a withdraw amount:"))
if(acc_bal>=wd_amnt):
    if(wd_amnt%100==0):
        print("transaction successful")
    else:
        print("enter amount in multiple of 100")
else:
    print("insufficient balance")
   
        

        

````

````python
code explantaion
1. It asks for the account balance (acc_bal) and the amount to withdraw (wd_amnt), 
then converts both inputs to decimal numbers with float().
2. if acc_bal >= wd_amnt: checks whether the balance covers the withdrawal.
3. If there is enough balance, if wd_amnt % 100 == 0: checks whether the withdrawal amount is an exact multiple of 100.
The % operator gives the remainder after division, so a remainder of 0 means it’s a multiple of 100.
4. It prints “transaction successful” if both checks pass; otherwise, it asks for an amount in multiples of 100.
5. If the balance is too low, it prints “insufficient balance.”
This checks the amount and balance, but it doesn’t update the balance after a successful withdrawal.

  
````

````python
5. discount calculator
take the total bill amount as input.

if bill >= 5000
if bill >= 10000 --> give 20% disount
else--> give 10% discount
else
if bill>=2000 --> give 5% discount
else
if bill >= 2000 --> give 5% discount
else--> "no discount"

````

````python
bill_amnt = float(input("Enter a balance amount: "))

if bill_amnt >= 2000:
    if bill_amnt >= 5000:
        if bill_amnt >= 10000:
            print("20% discount")
            disc_amnt = bill_amnt * 0.20
        else:
            print("10% discount")
            disc_amnt = bill_amnt * 0.10
    else:
        print("5% discount")
        disc_amnt = bill_amnt * 0.05
else:
    print("No discount")
    disc_amnt = 0

print("The discount amount is:", disc_amnt)
print("The bill amount to pay is:", bill_amnt - disc_amnt)
````

````python
code explanation
This code asks for a bill amount, calculates a discount based on the amount, and prints the discount and the final amount to pay:

- **₹10,000 or more:** 20% discount
- **₹5,000 to less than ₹10,000:** 10% discount
- **₹2,000 to less than ₹5,000:** 5% discount
- **Less than ₹2,000:** no discount

For example, if the bill is ₹6,000, the discount is ₹600, so the amount to pay is ₹5,400.
````

````python
a = float(input("Enter the first side: "))
b = float(input("Enter the second side: "))
c = float(input("Enter the third side: "))

if a + b > c and a + c > b and b + c > a:
    if a == b and b == c:
        print("Equilateral triangle")
    elif a == b or a == c or b == c:
        print("Isosceles triangle")
    else:
        print("Scalene triangle")
else:
    print("Invalid triangle")
````

````python

       Valid triangle: The sum of any two sides must be greater than the third side.
Equilateral: All three sides are equal.
Isosceles: Exactly two sides are equal.
Scalene: All three sides are different.
Invalid triangle: The sides fail the triangle rule. 
````

````python
account_type = input("Enter account type (savings/current): ").lower()
balance = float(input("Enter account balance: "))
transaction_type = input("Enter transaction type (deposit/withdrawal): ").lower()
amount = float(input("Enter transaction amount: "))

if amount <= 0:
    print("Enter a positive amount")
elif account_type == "savings":
    if transaction_type == "deposit":
        balance += amount
        print("Deposit successful")
    elif transaction_type == "withdrawal":
        if balance - amount >= 1000:
            balance -= amount
            print("Withdrawal successful")
        else:
            print("Withdrawal denied: minimum balance is ₹1,000")
    else:
        print("Invalid transaction type")
elif account_type == "current":
    if transaction_type == "deposit":
        balance += amount
        print("Deposit successful")
    elif transaction_type == "withdrawal":
        if balance - amount >= 5000:
            balance -= amount
            print("Withdrawal successful")
        else:
            print("Withdrawal denied: minimum balance is ₹5,000")
    else:
        print("Invalid transaction type")
else:
    print("Invalid account type")

print("Account balance:", balance)
````
